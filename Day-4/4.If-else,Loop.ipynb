```python
Temp=int(input("Enter Tempreture"))
if Temp<0:
    print("Freezing Weather")
elif Temp<10:
    print("Very Cold Weather")
elif Temp<20:
    print("Cold Weather")
elif Temp<30:
    print("Normal Weather")
else:
    print("Hot Weather")
```

```python
Temp=int(input("Enter Tempreture"))
if Temp>25:
  humidity=int(input("Enter Humidity"))
  if humidity>50:
    print("Hot and Humid")
  else:
    print("Only Hot Weather")
else:
    print("Cold Weather")
```
