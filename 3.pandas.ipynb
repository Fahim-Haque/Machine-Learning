## 1. Pandas Series
A Series is a one-dimensional labeled array.

```python
fruits = ['Mango', 'Banana', 'orange']
s_data = pd.Series(fruits, index=[1, 2, 3], name='Fruits')
s_data
```

## 2. DataFrame Creation
A DataFrame is a 2D table.

```python
df = pd.DataFrame({
    'Fruits': ['Mango', 'Apple', 'Banana', 'Orange'],
    'Quantity': [40, 20, 25, 10],
    'Price': [80, 100, 50, 70]
})
df
```

## 8. Dropping Columns

## 9. Combining Two Datasets

```python
df1 = pd.DataFrame({
    'Fruits': ['Mango', 'Banana', 'Grapes', 'Apple', 'Orange'],
    'Price': [60, 40, 75, 100, 65]
})
```

```python
df2 = pd.DataFrame({
    'Fruits': ['Apple', 'Orange', 'Papaya', 'Pineapple', 'Mango'],
    'Price': [120, 60, 30, 70, 50]
})
```

```python
df3 = pd.concat([df1, df2], axis=0, ignore_index=True)
df3
```
